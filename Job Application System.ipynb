# Reading Original Text from a file

```python
from PyPDF2 import PdfReader
import docx2txt
import os
from spire.doc import *
from spire.doc.common import *
from word_tokenizer import WordTokenizer
from gensim.models import Word2Vec
import numpy as np
```

Reading a PDF

```python
def readPDF(filepath):
    reader = PdfReader(filepath)
    number_of_pages = len(reader.pages)
    pages = reader.pages
    texts = []
    for page in pages:
        text = page.extract_text()
        texts.append(text)
    return texts
```

Reading a Word file

```python
def readWord(filepath):
    # extract text
    if (filepath.endswith('.docx')):
        text = docx2txt.process(filepath)
    elif (filepath.endswith('.doc')):
        document = Document()
        document.LoadFromFile(filepath)
        text = document.GetText().lstrip('evaluation warning: the document was created with spire.doc for python.')
    return text
```

Reading a txt file

```python
def readTxt(filepath):
    file = open(filepath, "r")
    texts = []
    for line in file:
        texts.append(line)
    return texts
```

# Extract Text from Sources

Function to parse the text content from sources

```python
def parseText(filepath):
    if (filepath.endswith(".pdf")):
        texts = readPDF(filepath)
        texts_str = ''
        for page in texts:
            texts_str += page.lower()
        return texts_str
    elif (filepath.endswith('.doc') or filepath.endswith('docx')):
        return readWord(filepath).lower()
    elif (filepath.endswith('.txt')):
        texts = readTxt(filepath)
        texts_str = ''
        for line in texts:
            texts_str += line.lower()
        return texts_str
    else:
        return None #invalid filepath or file format
```

# Naive Approch

Check how many times each word overlaps from one document to another

Preprocessing

Text Normalization - Removing special characters and punctuations from texts 

```python
from string import punctuation
delim = punctuation
```

```python
resumes = [resume.replace("\t", "").replace("\n", "").replace("\r", "") for resume in resumes]
cover_letters = [cover_letter.replace("\t", "").replace("\n", "").replace("\r", "") for cover_letter in cover_letters]
jobDescs = [jobDesc.replace("\t", "").replace("\n", "").replace("\r", "") for jobDesc in jobDescs]
```

Text Tokenization - Splitting texts by word

Split resume texts

```python
word_tokenizer = WordTokenizer()
resumes_splits = [word_tokenizer.tokenize(resume) for resume in resumes]
resumes_splits
```

Split Cover Letters Texts

```python
cover_letters_splits = [word_tokenizer.tokenize(cover_letter) for cover_letter in cover_letters]
(cover_letters_splits)
```

Split Job Descriptions Texts

```python
jobDescs_splits = [word_tokenizer.tokenize(jobDesc) for jobDesc in jobDescs]
jobDescs_splits
```

Find occurences of each word 
(by comparing the resume + cover letter, with job description)

```python
def getOccurences(targetText_split, sourceText1_split, sourceText2_split):
    occurences = {}
    for keyword in targetText_split:
        keyword = keyword.translate(str.maketrans('', '', punctuation))
        keyword_countInSrc1 = sourceText1_split.count(keyword)
        keyword_countInSrc2 = sourceText2_split.count(keyword)
        occurences.update({
            keyword: keyword_countInSrc1 + keyword_countInSrc2
        })
    return occurences
```

```python
occurences = []
min_length = min(len(jobDescs_splits), len(resumes_splits), len(cover_letters_splits))
for i in range(min_length):
    occurences.append(getOccurences(jobDescs_splits[i], resumes_splits[i], cover_letters_splits[i]))
occurences
```

```python
# Similarity Score
similarity_score = []
for occurence in occurences:
    similarity_score.append(sum(occurence.values()))
similarity_score
```

Calculate similarity values for each resume + cover letter + job description set, by averaging the occurence numbers of each word

```python
def calculateSimilarity(occurences_source, occurences_target, targetText_split):
    totalNumOfWords_jobDesc = len(targetText_split) #number of words in target text
    valueSum = 0
    for occurence_source in occurences_source:
        key = occurence_source
        value = occurences_source[occurence_source]
        if (key in occurences_target):
            if (value == occurences_target[key]):
                valueSum += value
        else:
            continue
        if (value > 1):
            diff = (value - 1)
            totalNumOfWords_jobDesc += diff #total number of words should be enlarged accordingly
            valueSum += diff
        elif (value == 1):
            valueSum += 1
        else:
            continue
    return valueSum / totalNumOfWords_jobDesc #similarity (ranging from 0 to 1)
```

```python
for i in range(len(occurences)):
    currSimilarityScore = str(calculateSimilarity(occurences[i], occurences[i], jobDescs_splits[i]))
    print(f"Similarity Score for Set {i+1} job application: {currSimilarityScore}")
```

Conclusion: Under the Naive Approach, Set 1 Job Application has a submitted resume and cover letter the most similar to the job description, comparing to other sets of Job Applications, because its resume and cover letter have the greatest number of words matching the job description. This Job Application should therefore be the most likely to bypass any Job Application System scanning.

# Similarity Ratio

Assumption:
This approach allows the library calculate text diff itself (ie. everything is black-boxed).
It is obvious that this method is not suitable to compare application texts (ie. resume text + cover letter text) with Job Description, maybe because the application text is a concatenated string from 2 different sources.

```python
from difflib import SequenceMatcher
min_length = min(len(resumes), len(cover_letters), len(jobDescs))
for i in range(min_length):
    similarity_ratio = SequenceMatcher(None, resumes[i] + cover_letters[i], jobDescs[i]).ratio()
    print(f"Similarity ratio using SequenceMatcher in Set {i+1} job application: {similarity_ratio}")
```

In different comparing order

```python
from difflib import SequenceMatcher
for i in range(min_length):
    similarity_ratio = SequenceMatcher(None, jobDescs[i], resumes[i] + cover_letters[i]).ratio()
    print(f"Similarity ratio using SequenceMatcher in Set {i+1} job application: {similarity_ratio}")
```

Conclusion: By using the Similarity Ratio, in any order, it looks like Set 2 job application has the highest similarity ratio, comparing to others. 

# Further Discussion about Similarity Ratio

Deduction 1: Since this approach returns a similarity score far lower than that calculated from my "Naive Approach", and because my "Naive approach" calculates the score based on each complete English word without associated special characters, we know that the "Similarity Ratio" might not even rely on text semantics (ie. meaning of the passage) to evaluate similarities. 
Instead, it might be relying on characters comparison.

Deduction 2: When texts are being put in different orders, the similarity ratios are different. We can deduce that it probably uses mathematical division to calculate matching characters.

```python
def calculateSimilarityByCharacter(text1, text2, compareText):
    occurences_fromSource = {}
    occurences_toCompare = {}
    
    # extract characters and occurences
    # resume
    for c in text1:
        if (c in occurences_fromSource):
            occurences_fromSource.update({
                c: occurences_fromSource[c]+ 1
            })
        else:
            occurences_fromSource.update({
                c: 1
            })
    # cover letter
    for c in text2:
        if (c in occurences_fromSource):
            occurences_fromSource.update({
                c: occurences_fromSource[c]+ 1
            })
        else:
            occurences_fromSource.update({
                c: 1
            })
    # job description
    for c in compareText:
        if (c in occurences_toCompare):
            occurences_toCompare.update({
                c: occurences_toCompare[c]+ 1
            })
        else:
            occurences_toCompare.update({
                c: 1
            })
    
    #finding number of matching characters
    similarity = 0
    matchingChars = 0
    totalCharacters_ToCompare = len(occurences_toCompare)
    totalCharacters_FromSource = len(occurences_fromSource)
    #matching characters to Job Description
    for occurence in occurences_toCompare:
        if (occurence in occurences_fromSource):
            matchingChars += occurences_fromSource[occurence]
    #calculate similarity
    similarity = (2 * matchingChars) / (totalCharacters_ToCompare + totalCharacters_FromSource)
    if (abs(similarity) > 1):
        similarity = pow(similarity, -1) #invert fraction
    return abs(similarity)
```

```python
min_length = min(len(resumes), len(cover_letters), len(jobDescs))
for i in range(min_length):
    similarity = calculateSimilarityByCharacter(resumes[i], cover_letters[i], jobDescs[i])
    print(f"Similarity Score in Set {i+1} job application: {str(similarity)}")
```

Conclusion: From the above results, since this similarity score is quite close to the ratios in both orderings that we used with SequenceMatcher (to calculate the similarity ratio based on character diff), it is affirmative that our deductions are correct:
1. Similarity Ratio is based on character comparison.
2. Such method uses mathematical division to calculate the similarity score with the number of matching characters.

Conclusion: Again, by using the Similarity Ratio, it looks like Set 1 job application has the highest similarity ratio/score, comparing to others.

# Extension to Similarity Ratio: get_opcodes

Principle behind is to evaluate the similarity score by determining how many operations needed to transform the original source to the target text for comparison.
The greater the number is, the original text is less similar to the target text.

```python
from difflib import SequenceMatcher
min_length = min(len(resumes), len(cover_letters), len(jobDescs))
for i in range(min_length):
    totalTransformation = len(SequenceMatcher(None, resumes[i] + cover_letters[i], jobDescs[i]).get_opcodes())
    print(f"In Set {i+1} job application, it takes {totalTransformation} transformations to change the source text to target text.")
```

Under this approach, it looks like in Set 1 Job Application the Resume + Cover Letter is more similar to the Job Description, than those in Set 2 Job Application, since it takes less transformations to make both text the same.

```python
from collections import Counter

min_length = min(len(resumes), len(cover_letters), len(jobDescs))

for i in range(min_length):
    common = Counter((resumes[i] + cover_letters[i]).split()) & Counter(jobDescs[i].split())
    print(common)
    similarity = calculateSimilarity(common, common, jobDescs_splits[i])
    print(f"Similarity Score of Set {i+1} Job Application: {similarity}")
```

# Cosine Similarity

```python
# Combine all resumes into a single list of words
all_resumes = [word for resume in resumes_splits for word in resume]
print(f"Length of all resumes: {len(all_resumes)}")
```

```python
# Train Word2Vec model
model = Word2Vec(sentences=[all_resumes], vector_size=100, window=5, min_count=1, workers=4)
model
```

```python
def get_vector(text, model):
    vector = [model.wv[word] for word in text if word in model.wv]
    return sum(vector) / len(vector)
```

```python
# Convert each resume into a vector by averaging the word vectors
resume_vectors = np.array([get_vector(resume, model) for resume in resumes_splits])
resume_vectors.shape
```

There are 3 Resumes, each with 100 features vectorized.

```python
# Convert each cover letter into a vector by averaging the word vectors
cover_letter_vectors = np.array([get_vector(cover_letter, model) for cover_letter in cover_letters_splits])
cover_letter_vectors.shape
```

There are 4 Cover Letters, each with 100 features vectorized. 

```python
# Convert each job description into a vector by averaging the word vectors
job_desc_vectors = np.array([get_vector(job_desc, model) for job_desc in jobDescs_splits])
job_desc_vectors.shape
```

There are 4 job descriptions, each with 100 features vectorized.

```python
# Calculate Cosine Similarity
def cosine_similarity(resume, cover_letter, job_desc):
    sourceText = resume + cover_letter
    text2Compare = job_desc
    similarity = np.dot(sourceText, text2Compare) / (np.linalg.norm(sourceText) * np.linalg.norm(text2Compare))
    return similarity
```

```python
for i in range(min_length):
    similarity = cosine_similarity(resume_vectors[i], cover_letter_vectors[i], job_desc_vectors[i])
    print(f"Cosine Similarity for Set {i+1} job application: {similarity}")
cosine_similarity(resume_vectors[0], cover_letter_vectors[0], job_desc_vectors[0])
```

Cosine Similarity estimates scores overly high, which might be a cause of overfitting. 
Transformation-based approach estimates scores reasonably in the middle, with consistent conclusions drawn. 
Similarity ratio estimates scores overly low, which might be a cause of underfitting. 
The naiive counting-based approach also estimmates scores reasonably, slightly higher than that in transformation-based approaches, but conclusions are inconsistent.
